```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Set Seaborn style
sns.set(style="whitegrid", context="talk", font_scale=1.1)

# Set figure size
plt.figure(figsize=(8, 6))

# Updated datasets
datasets = ['Retailing', 'ICUstays', 'Adult Income','Purchasing', 'MIMIC-III (Avg)']
models = ['GPT-4', 'DeepSeek-Chat']

# Updated F1 Scores (example values; adjust as needed)
f1_scores = np.array([
    [1.00, 1.00],     # Retailing
    [1.00, 1.00],     # Transfers
    [0.97, 1.00],      # Adult Income
    [0.93, 0.99],     # Purchasing
    [0.95, 0.98]     # MIMIC-III (Avg)
    
])

# Optional: precision and recall
precision = np.array([
    [1.00, 1.00],
    [1.00, 1.00],
    [0.94, 1.00],
    [0.92, 0.91],
    [1.00, 1.00]
])

recall = np.array([
    [1.00, 1.00],
    [1.00, 1.00],
    [0.92, 0.94],
    [0.90, 0.89],
    [0.93, 0.91]
])

# Color palette and bar plot configuration
colors = sns.color_palette("Set2", len(models))
bar_width = 0.35
index = np.arange(len(datasets))

# Bar plot
for i, model in enumerate(models):
    plt.bar(index + i * bar_width, f1_scores[:, i], bar_width,
            label=model, alpha=0.9, color=colors[i], edgecolor='black')

plt.xlabel('Dataset')
plt.ylabel('F1 Score')
plt.xticks(index + bar_width / 2, datasets, rotation=20)
plt.ylim(0.85, 1.05)
plt.legend(loc='upper right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()



```

```python
import numpy as np
import matplotlib.pyplot as plt

# Set consistent style
plt.style.use('seaborn-v0_8-whitegrid')
colors = ['#1f77b4', '#ff7f0e']  # Blue for GPT-4, Orange for DeepSeek

# Create figure with constrained layout
fig, ax = plt.subplots(figsize=(8, 6), dpi=300)

# Data - now properly aligned with 10 temperature points
temperatures = np.linspace(0.1, 1.0, 10)
gpt4_f1 = [0.82, 0.83, 0.84, 0.85, 0.84, 0.83, 0.82, 0.81, 0.80, 0.79]  # Example realistic trend
deepseek_f1 = np.linspace(0.79, 0.77, 10) + np.random.normal(0, 0.004, 10)

# Plot lines with improved markers and styling
ax.plot(temperatures, gpt4_f1, 
        marker='o', markersize=8, linestyle='-', linewidth=2.5,
        color=colors[0], label='GPT-4', alpha=0.9)

ax.plot(temperatures, deepseek_f1, 
        marker='s', markersize=7, linestyle='-', linewidth=2.5, 
        color=colors[1], label='DeepSeek-Chat', alpha=0.9)

# Axis labels with improved font
ax.set_xlabel('Temperature (τ)', fontsize=12, labelpad=10)
ax.set_ylabel('F1 Score', fontsize=12, labelpad=10)

# Axis limits and ticks
ax.set_ylim(0.75, 0.86)
ax.set_yticks(np.arange(0.75, 0.87, 0.02))
ax.tick_params(axis='both', which='major', labelsize=10)

# Grid and legend
ax.grid(True, linestyle='--', alpha=0.6)
ax.legend(loc='upper right', framealpha=1, fontsize=11)

# Remove top/right spines
ax.spines[['top', 'right']].set_visible(False)

# Save with tight layout and high resolution
plt.tight_layout(pad=1.5)
plt.savefig('logic_reasoning_temperature_effects.png', 
           dpi=300, bbox_inches='tight', transparent=False)
plt.show()
```
